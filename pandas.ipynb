pandas indexinf and selection

```python
data= {
    "Name" : ["Ali", "Rehman", "raza", "Farhan"],
    "Ages" : [10,20,16,17],
    "class":[1,2,3,4],
    "City": ["skardu", "bahawalpur","Lahore","karachi"],

}

df = pd.DataFrame(data)
print(df)
```

```python
subset_loc = df.loc[1:3, ["Name", "City"]]
print(subset_loc)
                    
```

```python
subset_iloc= df.iloc[1:3 , 1:3]
print(subset_iloc)
```

```python
high_salary= df[df["Ages"]> 15]
print(high_salary)
```

```python
df.loc[df["Ages"] <20, "'class"]= 6
print(df.loc[df["Ages"] <20, "'class"])
```
