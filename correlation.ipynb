```python
import pandas as pd
import numpy as np

data = pd.Series([10, 12, 11, 13, 12, 100])

mean = data.mean()
std = data.std(ddof=0)

z_score = (data - mean) / std

outliers = data[abs(z_score) > 3]

print("Mean:", mean)
print("Z-Scores:")
print(z_score)
print("Outliers:")
print(outliers)
```

```python
import pandas as pd

data = pd.Series([10, 12, 11, 13, 12, 100])

Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[(data < lower) | (data > upper)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Boundary:", lower)
print("Upper Boundary:", upper)
print("Outliers:", outliers.tolist())
```
