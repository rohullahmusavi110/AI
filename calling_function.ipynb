Function and parameters

```python
def add(a,b):
    return a+b
    
```

```python
result=add(10,15)
print(result)
```

```python
def square(number):
    return(number* number)
res=square(10)
print(res)
```

lambda Function

```python
add= lambda a,b : a+b
result=add(10,20)
print(result)
```

```python
nums =[1,2,3,4,5]
squared_num=list(map(lambda x: x*x, nums))
squared_num
```

```python
num=[2,3,4,8]
div=list(map(lambda x: x%2==0 , num))
print(div)
```
