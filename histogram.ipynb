```python
import matplotlib.pyplot as plt

morning = [45, 50, 55, 60, 65, 70, 75, 80]
evening = [50, 55, 60, 65, 70, 75, 80, 90]

plt.hist(
    morning,
    bins=5,
    alpha=0.6,
    label="Morning Students"
)

plt.hist(
    evening,
    bins=5,
    alpha=0.6,
    label="Evening Students"
)

plt.title("Morning vs Evening Students")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 200, 180, 300]

plt.figure(figsize=(9, 5))

plt.plot(
    months,
    sales,
    marker="o",
    linewidth=2,
    label="Monthly Sales"
)

plt.title("Monthly Sales Report", fontsize=16)
plt.xlabel("Months", fontsize=12)
plt.ylabel("Sales", fontsize=12)

plt.xticks(rotation=30)
plt.yticks(rotation=90)

plt.grid(True, linestyle="--", alpha=0.5)

plt.legend(loc="upper left")

plt.annotate(
    "Highest Sales",
    xy=("May", 300),
    xytext=("Apr", 260),
    arrowprops=dict(arrowstyle="->")
)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# Sample data
months = ["January", "February", "March", "April", "May", "June"]
sales = [100, 150, 200, 180, 250, 350]
profit = [20, 40, 60, 50, 80, 120]

# Figure
plt.figure(figsize=(10, 6))

# Plot two lines
plt.plot(months, sales, marker="o", label="Sales")
plt.plot(months, profit, marker="s", label="Profit")

# Title and labels
plt.title("Monthly Business Performance", fontsize=16)
plt.xlabel("Months")
plt.ylabel("Amount")

# Tick customization
plt.xticks(rotation=30)
plt.gca().yaxis.set_major_locator(
    ticker.MultipleLocator(50)
)

# Legend
plt.legend(loc="upper left")

# Grid
plt.grid(alpha=0.3)

# Annotation
plt.annotate(
    "Highest Sales",
    xy=("June", 350),
    xytext=("April", 320),
    arrowprops=dict(arrowstyle="->")
)

# Layout
plt.tight_layout()

# Export
plt.savefig(
    "business_report.png",
    dpi=300,
    bbox_inches="tight"
)

# Display
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 200, 180]

# Error values
errors = [10, 15, 20, 12]

plt.figure(figsize=(7,10))
plt.errorbar(
    months,
    sales,
    yerr=errors,
    capsize=5,
    alpha=0.5,
    color="r",
    label="average sales"
)

plt.title("Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

mean = [100, 150, 200, 180, 220]

lower = [90, 135, 180, 165, 200]
upper = [110, 165, 220, 195, 240]

plt.figure(figsize=(9, 5))

plt.plot(months, mean, marker='o', label='Mean Sales')

plt.fill_between(
    months,
    lower,
    upper,
    alpha=0.3,
    label='Confidence Interval'
)

plt.title("Sales with Confidence Interval")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.legend()
plt.grid()
plt.show()
```
