```python
import matplotlib.pyplot as plt
days =[1,2,3,4,5,6,7]
sales=[200,522,363,789,928,963,1900]
plt.plot(days , sales,
    marker= "o",
    linewidth=2,
    alpha=0.8,
    label="Daily Sales")

plt.xlabel=days
plt.ylabel=sales
plt.title("daily sales Trend")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
student=["ali","Ahmed","Hassan","Raza"]
marks=[10,38,26,30]
plt.bar(student, marks, width=0.6)

plt.title("Student marks chart")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.ylim(0,100)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
student=["ali","Ahmed","Hassan","Raza"]
math=[10,38,26,30]
english=[23,19,20,26]

x= np.arange(len(student))
width=0.35
plt.bar(x - width/2 , math, width,label="Math")

plt.bar(x+ width/2 , english,width , label ="English")

plt.title("Student marks chart")
plt.xticks(x , student)
plt.xlabel("Student")
plt.ylabel("Marks")
plt.ylim(0,100)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

classes = ["Class 5", "Class 6", "Class 7", "Class 8"]
boys = [20, 25, 22, 28]
girls = [18, 20, 24, 17]

plt.bar(classes, boys, label="Boys")
plt.bar(classes, girls, bottom=boys, label="Girls")

plt.title("Class Strength")
plt.xlabel("Classes")
plt.ylabel("Number of Students")
plt.legend()

plt.show()
```
