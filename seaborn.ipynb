```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Load dataset
df = pd.read_csv("seaborn_pr.csv")

# Calculate correlation
corr = df.corr(numeric_only=True)

# Basic heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr)
plt.title("Basic Correlation Heatmap")
plt.show()

# Annotated heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)
plt.title("Annotated Correlation Heatmap")
plt.show()

# Masked heatmap
mask = np.triu(
    np.ones_like(corr, dtype=bool),
    k=1
)

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    mask=mask
)
plt.title("Masked Correlation Heatmap")
plt.show()

# Clustermap
sns.clustermap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    figsize=(8, 8)
)
plt.show()

# Top correlated pairs
upper = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
)

top_pairs = upper.stack().sort_values(
    key=abs,
    ascending=False
)

print("Top Correlated Feature Pairs:")
print(top_pairs)
```

```python
import numpy as np
marks= np.array([60,70,80,90,100])

mean=np.mean(marks)

std= np.std(marks)
#apply z score

z_score = (marks - mean) / std

print(mean , std)

print(z_score)
print("New Mean", np.mean(z_score))
print("New Standard deviation", np.std(z_score))
```

```python
import numpy as np

marks = np.array([60, 70, 80, 90, 100])

# Find minimum and maximum
minimum = np.min(marks)
maximum = np.max(marks)

# Apply Min-Max Scaling
scaled = (marks - minimum) / (maximum - minimum)

print("Original Data:", marks)
print("Minimum:", minimum)
print("Maximum:", maximum)
print("Scaled Data:", scaled)

print("New Minimum:", np.min(scaled))
print("New Maximum:", np.max(scaled))
```

```python
import matplotlib.pyplot as plt
import numpy as np

sales = np.array([100, 120, 150, 130, 140, 160, 180, 200, 5000])

log_sales = np.log1p(sales)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.hist(sales, bins=5)
plt.title("Original Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(log_sales, bins=5)
plt.title("After Log Transformation")
plt.xlabel("Log Sales")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

df = pd.DataFrame({
    "Order_Date": [
        "2026-01-15",
        "2026-02-20",
        "2026-03-10",
        "2026-04-25"
    ]
})

# Convert column into datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Extract date features
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day
df["Day_Name"] = df["Order_Date"].dt.day_name()
df["Quarter"] = df["Order_Date"].dt.quarter

print(df)
```
