Handling Exceptions

```python
try:
    with open("missing.txt", "r") as file2:
        print(file2.read())
except FileNotFoundError:
    print("The file does not exist.")
```

```python
filename ="my_info.txt"
name="My_doc"
age= 20
```

```python
try:
    with open(filename, "r") as new_file:
        print(new_file.read())
except FileNotFoundError:
    with open(filename, "w") as new_files:
        new_files.write(f" {name} \n {age}")
        with open(filename, "r") as new_file:
            print(new_files.read())
        


```

```python
try:
    num = int(input("Enter a number: "))
    print(10/num)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except ValueError:
    print("Error: Invalid input. Please enter a valid number.")
except Exception as e:
    print(f"An unexpected error occurred: {e}") 
```

```python
try:
    print("running code")
except:
    print("error")
else:
    print("code ran successfully")
finally:
    print("code ran successfully")
```

Raising errors / exception

```python
age=-8
if age<0:
    raise ValueError("Age cannot be negative.")
```

```python
class invalidAgeError(Exception):
    pass
```

```python
age=int(input("Enter your age: "))
if age < 0:
    raise invalidAgeError(f" {age} Age cannot be negative.")
```

```python
try:
    age=int(input("Enter your age: "))
    if age < 0:
        raise invalidAgeError(f" {age} Age cannot be negative.")
except invalidAgeError as e:
    print(f"Invalid age: {e}")
```
