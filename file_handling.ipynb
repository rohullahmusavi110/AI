Handling Exceptions

```python
try:
    with open("missing.txt", "r") as file2:
        print(file2.read())
except FileNotFoundError:
    print("The file does not exist.")
```

```python
filename ="my_info.txt"
name="My_doc"
age= 20
```

```python
try:
    with open(filename, "r") as new_file:
        print(new_file.read())
except FileNotFoundError:
    with open(filename, "w") as new_files:
        new_files.write(f" {name} \n {age}")
        with open(filename, "r") as new_file:
            print(new_files.read())
        


```
