```python
Customer = pd.DataFrame({
    "customer_id":[1,2,3],
    "Name":["Alice","Bob","Charlie"]
})



```

```python
orders = pd.DataFrame({
    "orders_id":[101,102,103,104],
    "customer_id":[1,2,2,3],
    "Amount": [100,120,150,200]
    
})
```

```python

df_inner= pd.merge(Customer , orders, on="customer_id", how="inner")
print(df_inner)
```

```python
df_left=pd.merge(Customer , orders, on="customer_id", how="left")
df_left
```

```python
df_outer=pd.merge(Customer , orders, on="customer_id", how="outer")
df_outer
```
