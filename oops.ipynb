```python
class student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age= age
        self.grade = grade

    def display_info(self):
        print(f"Student name: {self.name}")
        print(f"Student age: {self.age}")
        print(f"Student grade: {self.grade}")

    def is_eligible(self):
        if self.age >= 18:
            print(f"{self.name} is eligible for college") 
        else:
            print(f"{self.name} is not eligible for college")   
```
