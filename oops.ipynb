Encapsulation in python

```python
class student:
    def __init__(self, name , age):
        self.__age=age
        self.name=name

    def get_age(self):
        return self.__age
student=student("Ali", 17)
print(student.get_age(), student.name)
```

```python
class Student:
    
    def grade(self):
        return self.__grade
    def __init__(self, name,grade):
        self.__grade = grade

    def get_grade(self):
        return self.__grade
    def set_grade(self, grade):
        if 0<= grade <= 100:
            self.__grade=grade
        else:
            print("Invalid grade. Please enter a value between 0 and 100.")
    
```

```python
Student1=Student("Ali", 95)
Student1.set_grade(90)
Student1.set_grade(105)  
```

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs
    def __len__(self):
        return len(self.songs)
    def __add__(self, other):
        return Playlist(self.songs + other.songs)
```

```python
p1=Playlist(["Song 1", "Song 2", "Song 3"])
```

```python
p2=Playlist(["Song 4", "Song 5"])
```

```python
print(len(p1))  # Output: 3
print(len(p2))  # Output: 2
```

```python
class point:
    def __init__(self , x, y):
        self.x=x
        self.y=y
    def __str__(self):
        return f"Point({self.x}, {self.y})"
    def __add__(self, other):
        return point(self.x + other.x, self.y + other.y)
    
```

```python
p1=point(2, 3)
p2=point(4, 5)
```

```python
print(p1 + p2)  # Output: Point(6, 8)
```
